```julia
struct Item
    """
    name, 
    attack damage,
    ability power,
    magic resist,
    health,
    mana
    """
    name::String
    AH::Float64
    AP::Float64  
    MR::Float64
    H::Float64
    M::Float64
    PMS::Float64
    gold_value::Float64
    gold_efficiency::Float64
end

# Constructor function to compute gold_value automatically
function Item(name, gold_cost, AH, AP, MR, H, M, PMS; GpS=Dict("AH"=>50, "AP"=>20, "MR"=>20, "H"=>2.66667, "M"=>1, "PMS"=>53.0))
    gold_value = AH * GpS["AH"] + AP * GpS["AP"] + MR * GpS["MR"] + H * GpS["H"] + M * GpS["M"] + PMS * GpS["PMS"]
    gold_efficiency = gold_value / gold_cost
    return Item(name, AH, AP, MR, H, M, PMS, gold_value, gold_efficiency)
end
function print_gold_efficiency(item::Item)
    println("Gold Value of $(item.name): ", item.gold_value)
    println("Gold Efficiency of $(item.name): ", item.gold_efficiency)
    println(" ")
end
roa = Item("Rod of Ages unupgraded", 2600, 0.0, 45.0, 0.0, 350.0, 500.0, 0.0)
roa_upgraded = Item("Rod of Ages upgraded", 2600, 0.0, 75.0, 0.0, 450.0, 800.0, 0.0)

deathcap = Item("Rabadon's Deathcap", 3500, 0.0, 130.0, 0.0, 0.0, 0.0, 0.0)
deathcap_bonus = Item("Rabadon's Deathcap bonus", 3500, 0.0, (130.0)*1.3, 0.0, 0.0, 0.0, 0.0)

archangles_staff = Item("Archangel's Staff unupgraded", 2900, 25.0, 70.0, 0.0, 0.0, 600.0, 0.0)
seraphs_embrace = Item("Seraph's Embrace upgraded", 2900, 25.0, 70.0 + 0.02*1000.0, 0.0, 0.0, 1000.0, 0.0)

cosmic_drive = Item("Cosmic Drive", 3000, 25.0, 70.0, 0.0, 350.0, 0.0, 4.0)


print_gold_efficiency(roa)
print_gold_efficiency(roa_upgraded)
print_gold_efficiency(deathcap)
print_gold_efficiency(deathcap_bonus)
print_gold_efficiency(archangles_staff)
print_gold_efficiency(seraphs_embrace)
print_gold_efficiency(cosmic_drive)



```
